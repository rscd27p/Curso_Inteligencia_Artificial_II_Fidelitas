````python

# ============================================================
# EJEMPLOS DE AMBIGÜEDAD Y DE COMO RECORRER LISTAS
# ============================================================

frases = [
    "Qué buena nota",
    "Estoy muriéndome de hambre",
    "Buenísimo... se cayó el sistema otra vez",
    "Mae qué carga",
    "Excelente... perdimos todos los datos"
]

# Método 1 - usando enumerate

print("Método 1 - usando enumerate")
for i, frase in enumerate(frases):
    print(f"{i+1}. {frase}")

# Método 2 - Usando rango y contador

print("Método 2 - usando contador y tamaño de lista")
for i in range(len(frases)):
    print(f"{i+1}. {frases[i]}")

# Método 3 - Usando Iterador

print("Método 3 - Usando Iterador")
i = 0
for frase in frases:
  i=i+1
  print(f"{i}. {frase}")
````

- hojas de cálculo,
- o tablas SQL.

---

# Características de un DataFrame

Un DataFrame:

- organiza información en filas y columnas,
- permite analizar datos fácilmente,
- facilita limpieza y transformación de datos,
- y es ampliamente utilizado en:
  - Ciencia de Datos,
  - Machine Learning,
  - Deep Learning,
  - NLP,
  - y análisis estadístico.

---

````python
# ============================================================
# CREAR UN DATAFRAME
# ============================================================

# Importar pandas
import pandas as pd


# Crear datos de ejemplo utilizando un diccionario

# formato diccionarios
# {
#   key1 : value_1 # Llave son str
#   key2 : value_2
#   ....
#   keyn : value_n
# }
datos = {
    "Nombre": ["Ana", "Luis", "María", "Carlos", "Sofía"],
    "Edad": [22, 30, 27, 35, 24],
    "Ciudad": ["San José", "Heredia", "Cartago", "Alajuela", "Limón"],
    "Carrera": ["IA", "Software", "Ciberseguridad", "Redes", "IA"]
}
# Recuerden que todas las listas deben de ser del mismo tamaño
# Pueden partir de un diccionario

# Crear DataFrame
df = pd.DataFrame(datos)


# Mostrar DataFrame completo
df
````

````python
# ============================================================
# VISUALIZAR LAS PRIMERAS FILAS
# ============================================================

# head()
# ------------------------------------------------------------
# Muestra las primeras filas del DataFrame.
#
# Por defecto:
# muestra las primeras 5 filas.
#
# Muy útil para:
# - explorar datasets rápidamente
# - verificar estructura
# - inspeccionar datos
# ------------------------------------------------------------

df.head()
````

````python
# Mostrar solamente las primeras 3 filas

df.head(6)
````

````python
# ============================================================
# VISUALIZAR LAS ÚLTIMAS FILAS
# ============================================================

# tail()
# ------------------------------------------------------------
# Muestra las últimas filas del DataFrame.
#
# Muy útil para:
# - revisar datos finales
# - verificar cargas completas
# ------------------------------------------------------------

df.tail(2)
````

````python
# ============================================================
# INFORMACIÓN GENERAL DEL DATAFRAME
# ============================================================

# info()
# ------------------------------------------------------------
# Muestra información importante sobre:
# - columnas
# - tipos de datos
# - cantidad de valores
# - memoria utilizada
#
# Muy útil para:
# - análisis exploratorio
# - limpieza de datos
# ------------------------------------------------------------

df.info()
````

# SECCIÓN 5 — NLP Tradicional: Bag of Words
# ============================================================

Antes de transformers y LLMs, NLP dependía fuertemente de:

- reglas,
- estadísticas,
- ingeniería manual de características.

Uno de los enfoques más comunes era:
# Bag of Words

Bag of Words es una de las técnicas más clásicas de Procesamiento del Lenguaje Natural (PLN).

Su objetivo es:
# convertir texto en números.

Esto es necesario porque los modelos de Machine Learning NO pueden entender texto directamente.

---

# Idea Principal

Bag of Words analiza un conjunto de documentos y:

1. identifica las palabras únicas,
2. crea un vocabulario,
3. y cuenta cuántas veces aparece cada palabra.

---

# Importante

Bag of Words:

- NO entiende significado,
- NO entiende contexto,
- NO entiende sarcasmo,
- NO entiende semántica.

Solamente:
# cuenta palabras.

---

# Ejemplo Conceptual

Supongamos los siguientes documentos:

| Documento | Contenido |
|---|---|
| 1 | Me gustan los perros |
| 2 | Los perros son increíbles |
| 3 | Me gusta la pizza |

---

# ¿Qué hace Bag of Words?

Primero crea un vocabulario:

```text
gusta
gustan
increíbles
la
los
me
perros
pizza
son
```

Luego cuenta cuántas veces aparece cada palabra en cada documento.

---

# Resultado Simplificado

| Documento | perros | pizza | increíbles |
|---|---|---|---|
| 1 | 1 | 0 | 0 |
| 2 | 1 | 0 | 1 |
| 3 | 0 | 1 | 0 |

---

# Ventajas

Bag of Words:
- es simple,
- rápido,
- fácil de entender,
- y fue muy utilizado históricamente en NLP.

---

# Limitaciones

Bag of Words tiene problemas importantes:

## No entiende significado

Por ejemplo:

```text
carro
```

y

```text
automóvil
```

son similares para humanos.

Pero Bag of Words los ve como:
# palabras totalmente distintas.

---

# No entiende contexto

La frase:

```text
"Excelente... perdimos todos los datos"
```

podría parecer positiva porque contiene:
```text
Excelente
```

aunque realmente es sarcasmo.

---

# Objetivo del Ejercicio

En este ejercicio vamos a:

- convertir texto en números,
- visualizar el vocabulario,
- observar cómo funciona Bag of Words,
- y entender sus limitaciones.


````python
# ============================================================
# BAG OF WORDS
# ============================================================

# Lista de documentos de ejemplo
# ------------------------------------------------------------
# Cada elemento representa una oración o documento.
# ------------------------------------------------------------

from sklearn.feature_extraction.text import CountVectorizer

documentos = [
    "Me gustan los perros",
    "Los perros son increíbles",
    "Me gusta la pizza",
    "Saprissa el mejor de todos"
]


# Crear vectorizador Bag of Words
# ------------------------------------------------------------
# CountVectorizer:
# - identifica palabras únicas
# - crea un vocabulario
# - cuenta frecuencia de palabras
# ------------------------------------------------------------

vectorizador = CountVectorizer()


# Ajustar y transformar documentos
# ------------------------------------------------------------
# fit_transform():
#
# fit:
# aprende vocabulario
#
# transform:
# convierte texto en números
# ------------------------------------------------------------

X = vectorizador.fit_transform(documentos)


# Mostrar vocabulario detectado
# ------------------------------------------------------------
# get_feature_names_out():
# devuelve las palabras únicas encontradas.
# ------------------------------------------------------------

print(vectorizador.get_feature_names_out())
````

````python

# REPRESENTACIÓN NUMÉRICA

X.toarray()
````


# ¿Qué acaba de ocurrir?

El texto fue convertido a números.

Los modelos de Machine Learning NO entienden texto directamente.

Pero Bag of Words tiene limitaciones importantes:
- no entiende contexto,
- no entiende significado,
- ignora el orden de palabras,
- no comprende relaciones semánticas.

# ¿Qué significa `X.toarray()`?

Cuando utilizamos:

```python
X = vectorizador.fit_transform(documentos)
```

La variableX NO contiene una lista normal de Python.

En realidad contiene: una matriz dispersa (sparse matrix).

---

# ¿Qué es una matriz dispersa?

En NLP normalmente existen MUCHAS palabras posibles.

La mayoría de palabras:
- NO aparecen en cada documento.

Entonces la matriz contiene:
- muchísimos ceros.

Ejemplo:

| Documento | perros | pizza | increíbles |
|---|---|---|---|
| 1 | 1 | 0 | 0 |
| 2 | 1 | 0 | 1 |
| 3 | 0 | 1 | 0 |

Observe que:
- la mayoría de valores son 0.

---

# ¿Por qué usar matrices dispersas?

Porque ahorran:
- memoria,
- procesamiento,
- y almacenamiento.

Especialmente en datasets grandes.

---

# Entonces:
## ¿Qué hace `toarray()`?

Convierte la matriz dispersa en:
# una matriz normal de NumPy.

Esto permite visualizar los datos fácilmente.

---
